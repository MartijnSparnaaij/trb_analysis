```python
from pathlib import Path
from copy import deepcopy

import matplotlib.pyplot as plt
import matplotlib as mpl

from contact_plots import create_and_save_rel_close_contact_plot, create_and_save_rel_single_close_contact_plot
```

```python
data_filename = Path(r'C:\Users\martijnsparnaa\surfdrive\PhD\Activity scheduler\Simulations\2021_12\transfer\results_20211215\contact_data.npz')
```

```python
infra_names = [
    'base_case',
    'busy_corner',
    'long_restaurant']

infra_names_2_label = {
    'base_case': 'Base',
    'busy_corner': 'Busy', 
    'long_restaurant': 'Long'}

measures = (
    'nn_nts_bvd',
    'n_nts_bvd',
    'nn_ts_bvd',
    'nn_nts_svd',
    'nn_nts_lvd'    
)

interaction_type_labels = {'customer_customer':'Customer - Customer', 
                           'customer_staff':'Customer - Staff', 
                           'staff_customer':'Staff - Customer'}

measures_2_labels = {
    'nn_nts_bvd': 'Base',
    'n_nts_bvd': 'Neighbourhoods',
    'nn_ts_bvd': 'Time slots',
    'nn_nts_svd': 'Shorter visits',
    'nn_nts_lvd': 'Longer visits'
}
```

```python
divider_config = {
    'left':0.62,  
    'gap_hor':0,
    'right':0.05,
    'bottom':0.88,
    'gap_ver':0.0,
    'top':0.28,
    'v_legend':0.42,
    'share_y':True
}

main_ax_config = {'grid':{'visible':True, 'axis':'y'},  
                  'x_tick_labels':[measures_2_labels[measure] for measure in measures], 
                  'x_tick_labels_config':{'rotation':45, 'ha':'right'}}

bar_plot_config = {'width':0.8}

viridis = mpl.colormaps['viridis']

bar_color_order = [viridis(0.5), viridis(0.9), viridis(0.1)]


legend_config = {'loc':'upper center', 'bbox_to_anchor': (0.5, 1.1),'ncol':3, 'title':'Criterium', 'handlelength':1.2, 
                          'handleheight':0.5, 'title_fontsize':9, 'columnspacing':1.0, 'handletextpad':0.6}

v_line_config = {'linewidth':1.5, 'color':'0.6', 'linestyle':'--', 'zorder':2}

base_plot_config = {'measure_names':measures, 'main_ax_config':main_ax_config, 'v_line_config':v_line_config, 
                    'bar_plot_config':bar_plot_config, 'legend_config':legend_config, 'bar_color_order':bar_color_order,
                    'infra_names_2_label':infra_names_2_label}

fig_config_base = {'figsize':(6.5, 3.5), 'dpi':270}

save_pd = Path(r'C:\Users\martijnsparnaa\surfdrive\PhD\Graphics')
```

```python
%matplotlib wx

plt.style.use('tex_image_style.mplstyle')

interaction_type = 'customer_customer'

save_filename = save_pd.joinpath(f'{interaction_type}_rel_close_contacts.pdf')

fig_config = deepcopy(fig_config_base)
fig_config.update({'title': f'Close contacts per agent - {interaction_type}'})

plot_config = deepcopy(base_plot_config)
plot_config['interaction_type'] = interaction_type
plot_config['t_label'] = 'nc_dist=1.0_time=1.0_c_dist=1.5_time=15.0'
plots_config = []

for ii, infra_name in enumerate(infra_names):
    plot_config_sub = deepcopy(plot_config)
    plot_config_sub['infra_name'] = infra_name
    plot_config_sub['main_ax_config']['title'] = f'({chr(97+ii)}) {infra_names_2_label[infra_name]}'
    if ii == 0:
        plot_config_sub['main_ax_config']['y_label'] = '\# Close contacts per agent [-]'
    else:
        plot_config_sub['main_ax_config']['y_ticks_visible'] = False 
    
    plots_config.append(plot_config_sub)
    
create_and_save_rel_close_contact_plot(fig_config, divider_config, plots_config, data_filename, save_filename)
```

```python
%matplotlib wx
interaction_type = 'staff_customer'

save_filename = save_pd.joinpath(f'{interaction_type}_rel_close_contacts.pdf')

fig_config = deepcopy(fig_config_base)
fig_config.update({'title': f'Close contacts per agent - {interaction_type}'})

plot_config = deepcopy(base_plot_config)
plot_config['interaction_type'] = interaction_type
plot_config['t_label'] = 'nc_dist=1.0_time=1.0_c_dist=1.5_time=15.0'
plot_config['main_ax_config']['y_lim'] = [0,17.5]
plots_config = []

for ii, infra_name in enumerate(infra_names):
    plot_config_sub = deepcopy(plot_config)
    plot_config_sub['infra_name'] = infra_name
    plot_config_sub['main_ax_config']['title'] = f'({chr(97+ii)}) {infra_names_2_label[infra_name]}'
    if ii == 0:
        plot_config_sub['main_ax_config']['y_label'] = '\# Close contacts per agent [-]'
    else:
        plot_config_sub['main_ax_config']['y_ticks_visible'] = False
    
    plots_config.append(plot_config_sub)
    
create_and_save_rel_close_contact_plot(fig_config, divider_config, plots_config, data_filename, save_filename)
```

```python
%matplotlib wx
plt.style.use('tex_image_style.mplstyle')

divider_config_cust = deepcopy(divider_config)
divider_config_cust['left'] = 0.6
divider_config_cust['gap_hor'] = 0.6
divider_config_cust['share_y'] = False
divider_config_cust['top'] = 0.45

save_filename = save_pd.joinpath(f'staff_customer_staff_rel_close_contacts.pdf')

fig_config = deepcopy(fig_config_base)
fig_config.update({'title': f'Close contacts per agent - staff_customer_staff'})

blues = mpl.colormaps['Blues']
reds = mpl.colormaps['Reds']
greens = mpl.colormaps['Greens']

bar_color_order = [
    [blues(0.6), blues(0.35), blues(0.9)],
    [reds(0.65), reds(0.45), reds(0.95)],
    [greens(0.55), greens(0.3), greens(0.85)]
]

plot_config = deepcopy(base_plot_config)
plot_config['t_label'] = 'nc_dist=1.0_time=1.0_c_dist=1.5_time=15.0'
plot_config['bar_color_order'] = bar_color_order
plot_config['legend_config'].pop('title')
plot_config['main_ax_config']['x_lim'] = [-0.2,13]

plots_config = []

legend_labels = {
        'cc_count_both': 'Both',
        'cc_count_nc': '1 min. within 1m',
        'cc_count_c': '15 min. within 1.5m'
}

for ii, interaction_type in enumerate(('customer_staff', 'staff_customer')):
    plot_config_sub = deepcopy(plot_config)
    plot_config_sub['interaction_type'] = interaction_type
    sub_groups = []
    for infra_name in infra_names:
        sub_group = {'infra_name':infra_name,
                    'label': infra_names_2_label[infra_name],
                    'bar_values':['cc_count_both','cc_count_nc','cc_count_c'],
                    'legend_labels':legend_labels}
        sub_groups.append(sub_group)    
    plot_config_sub['sub_groups'] = sub_groups
    plot_config_sub['main_ax_config']['title'] = f'({chr(97+ii)}) {interaction_type_labels[interaction_type]}'
    plot_config_sub['main_ax_config']['y_label'] = 'Average \# contacts per agent [-]'
    if interaction_type == 'staff_customer':
        plot_config_sub['main_ax_config']['y_lim'] = [0,17.5]
    plots_config.append(plot_config_sub)
    
create_and_save_rel_single_close_contact_plot(fig_config, divider_config_cust, plots_config, data_filename, save_filename)
```
