```python
from pathlib import Path
from copy import deepcopy

import matplotlib.pyplot as plt
import matplotlib as mpl

from contact_plots import create_and_save_change_contact_plot, create_and_save_rel_change_contact_plot
```

```python
data_filename = Path(r'C:\Users\martijnsparnaa\surfdrive\PhD\Activity scheduler\Simulations\2021_12\transfer\results_20211215\contact_data.npz')
```

```python
infra_names = [
    'base_case',
    'busy_corner',
    'long_restaurant']

infra_names_2_label = {
    'base_case': 'Base case',
    'busy_corner': 'Busy corner', 
    'long_restaurant': 'Long restaurant'}

measures = (
    'nn_nts_bvd',
    'n_nts_bvd',
    'nn_ts_bvd',
    'nn_nts_svd',
    'nn_nts_lvd'    
)

measures_2_labels = {
    'nn_nts_bvd': 'Base',
    'n_nts_bvd': 'Neighbourhoods',
    'nn_ts_bvd': 'Time slots',
    'nn_nts_svd': 'Shorter visits',
    'nn_nts_lvd': 'Longer visits'
}

base_case = 'nc_dist=1.0_time=1.0_c_dist=1.5_time=15.0'

other_cases = [
    'nc_dist=1.0_time=1.0_c_dist=1.5_time=10.0', 
    'nc_dist=1.0_time=1.0_c_dist=1.5_time=5.0',
    'nc_dist=1.0_time=1.0_c_dist=1.8_time=15.0', 
    'nc_dist=1.0_time=1.0_c_dist=2.0_time=15.0',
    'nc_dist=1.0_time=1.0_c_dist=1.8_time=10.0', 
    'nc_dist=1.0_time=1.0_c_dist=1.8_time=5.0', 
    'nc_dist=1.0_time=1.0_c_dist=2.0_time=10.0', 
    'nc_dist=1.0_time=1.0_c_dist=2.0_time=5.0'
]

case_2_label = {
    'nc_dist=1.0_time=1.0_c_dist=1.5_time=10.0':'dist = 1.5m, time = 10 min', 
    'nc_dist=1.0_time=1.0_c_dist=1.5_time=5.0':'dist = 1.5m, time = 5 min', 
    'nc_dist=1.0_time=1.0_c_dist=1.8_time=15.0':'dist = 1.8m, time = 15 min', 
    'nc_dist=1.0_time=1.0_c_dist=2.0_time=15.0':'dist = 2.0m, time = 15 min', 
    'nc_dist=1.0_time=1.0_c_dist=1.8_time=10.0':'dist = 1.8m, time = 10 min', 
    'nc_dist=1.0_time=1.0_c_dist=1.8_time=5.0':'dist = 1.8m, time = 5 min', 
    'nc_dist=1.0_time=1.0_c_dist=2.0_time=10.0':'dist = 2.0m, time = 10 min', 
    'nc_dist=1.0_time=1.0_c_dist=2.0_time=5.0':'dist = 2.0m, time = 5 min'
}
```

```python
divider_config = {
    'left':0.75,  
    'gap_hor':0,
    'right':0.1,
    'bottom':1.05,
    'gap_ver':0.0,
    'top':0.3,
    'v_legend':0.70,
    'share_y':True
}

main_ax_config = {'grid':{'visible':True, 'axis':'y'},
                  'x_tick_labels':[measures_2_labels[measure] for measure in measures], 
                  'x_tick_labels_config':{'rotation':45, 'ha':'right'},
                  'x_lim':[-0.5, len(measures)-0.5]}

line_config = {'zorder':3}

viridis = mpl.colormaps['viridis']

color_step = 0.7/(len(other_cases) - 1)

color_order = [viridis(0.2+ii*color_step) for ii in range(len(other_cases))]
style_order = ['-', '--', '-', '--', '-', '--', ':', '-.']
width_order = [1.5]*8#[3, 2.5, 2, 1.5, 3, 2.5, 2, 1.5]
marker_order = ['s', 'X', 'o', '*', 'v', '^', '<', '>']

legend_config = {'loc':'upper center', 'ncol':3, 'handlelength':2, 
                          'handleheight':0.7, 'columnspacing':1.0, 'handletextpad':0.6}

v_line_config = {'linewidth':1.5, 'color':'0.6', 'linestyle':'--', 'zorder':2}

base_plot_config = {'measure_names':measures, 'main_ax_config':main_ax_config, 'v_line_config':v_line_config, 
                    'line_config':line_config, 'legend_config':legend_config, 'base_case':base_case,
                   'other_cases':other_cases, 'color_order':color_order, 'style_order':style_order, 
                    'width_order':width_order, 'marker_order':marker_order, 'case_2_label':case_2_label}

fig_config_base = {'figsize':(6.5, 4)}

save_pd = Path(r'C:\Users\martijnsparnaa\surfdrive\PhD\Graphics')
```

```python
%matplotlib wx

plt.style.use('tex_image_style.mplstyle')

interaction_type = 'customer_customer'

save_filename = save_pd.joinpath(f'{interaction_type}_change_close_contacts.pdf')

fig_config = deepcopy(fig_config_base)
fig_config.update({'title': f'Close contact change per agent - {interaction_type}'})

plot_config = deepcopy(base_plot_config)
plot_config['interaction_type'] = interaction_type
plots_config = []

for ii, infra_name in enumerate(infra_names):
    plot_config_sub = deepcopy(plot_config)
    plot_config_sub['infra_name'] = infra_name
    plot_config_sub['main_ax_config']['title'] = f'({chr(97+ii)}) {infra_names_2_label[infra_name]}'
    if ii == 0:
        plot_config_sub['main_ax_config']['y_label'] = 'Change in \# close contacts\n per agent [-]'
    else:
        plot_config_sub['main_ax_config']['y_ticks_visible'] = False 
    
    plots_config.append(plot_config_sub)
    
create_and_save_change_contact_plot(fig_config, divider_config, plots_config, data_filename, save_filename)
```
