```python
from pathlib import Path
import numpy as np

from contact_plots import get_data_from_files, get_close_contact_count_per_agent, get_contact_count_per_agent, load_data_from_file
```

```python
data_filename = Path(r'C:\Users\martijnsparnaa\surfdrive\PhD\Activity scheduler\Simulations\2021_12\transfer\results_20211215\contact_data.npz')
data_in_file = load_data_from_file(data_filename)

data_folder = Path(r'C:\Users\martijnsparnaa\surfdrive\PhD\Activity scheduler\Simulations\2021_12\transfer\results_20211215')
contact_data_list, infra_and_measure_name_2_data = get_data_from_files(data_folder, convert_2_minutes=True)
```

```python
infra_names = [
    'base_case',
    'busy_corner',
    'long_restaurant']

measures = (
    'nn_nts_bvd',
    'n_nts_bvd',
    'nn_ts_bvd',
    'nn_nts_svd',
    'nn_nts_lvd'    
)
```

```python
interaction_type = 'staff_customer'
t_label = 'nc_dist=1.0_time=1.0_c_dist=1.5_time=15.0'

for infra_name in infra_names:
    for measure_name in measures:
        contacts_per_agent = get_close_contact_count_per_agent(data_in_file, infra_name, measure_name, interaction_type, t_label)
        median = np.quantile(contacts_per_agent, 0.5)
        Q1 = np.quantile(contacts_per_agent, 0.25)
        Q3 = np.quantile(contacts_per_agent, 0.75)
        min_val = np.min(contacts_per_agent)
        max_val = np.max(contacts_per_agent)
        print(f'{infra_name} - {measure_name}: Med = {median}, Q1 = {Q1}, Q3 = {Q3}, IQR = {Q3-Q1:.3f}, max_diff = {max_val - min_val}, med_min_diff = {median-min_val}, med_max_diff = {max_val - median}')
```
