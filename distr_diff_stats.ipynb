```python
from pathlib import Path
import numpy as np

from contact_plots import get_data_from_files, print_diff_stats_table_base, CONTINUOUS, CATEGORICAL, INFRA_COMP, SETUP_COMP, CC_DEF_COMP, get_close_contact_distr, get_contact_distr
```

```python
data_filename = Path(r'C:\Users\martijnsparnaa\surfdrive\PhD\Activity scheduler\Simulations\2021_12\transfer\results_20211215\contact_data.npz')
close_contact_data = np.load(data_filename, allow_pickle=True)

data_folder = Path(r'C:\Users\martijnsparnaa\surfdrive\PhD\Activity scheduler\Simulations\2021_12\transfer\results_20211215')
_, contact_data = get_data_from_files(data_folder, convert_2_minutes=True)

```

```python
infra_names = [
    'base_case',
    'busy_corner',
    'long_restaurant']

measures = (
    'nn_nts_bvd',
    'n_nts_bvd',
    'nn_ts_bvd',
    'nn_nts_svd',
    'nn_nts_lvd'    
)

measures_2_labels = {
    'nn_nts_bvd': 'Base',
    'n_nts_bvd': 'Neighbourhoods',
    'nn_ts_bvd': 'Time_slots',
    'nn_nts_svd': 'Shorter_visits',
    'nn_nts_lvd': 'Longer_visits'
}

base_case_diff = 'nc_dist=1.0_time=1.0_c_dist=1.5_time=15.0'

other_cases_diff = [
    'nc_dist=1.0_time=1.0_c_dist=1.5_time=10.0', 
    'nc_dist=1.0_time=1.0_c_dist=1.5_time=5.0',
    'nc_dist=1.0_time=1.0_c_dist=1.8_time=15.0', 
    'nc_dist=1.0_time=1.0_c_dist=2.0_time=15.0',
    'nc_dist=1.0_time=1.0_c_dist=1.8_time=10.0', 
    'nc_dist=1.0_time=1.0_c_dist=1.8_time=5.0', 
    'nc_dist=1.0_time=1.0_c_dist=2.0_time=10.0', 
    'nc_dist=1.0_time=1.0_c_dist=2.0_time=5.0',
    'nc_None_c_dist=1.5_time=15.0'
]

t_label = 'nc_dist=1.0_time=1.0_c_dist=1.5_time=15.0'
```
