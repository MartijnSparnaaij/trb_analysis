```python
from pathlib import Path
from copy import deepcopy

import matplotlib.pyplot as plt
import matplotlib as mpl

from contact_plots import create_and_save_cum_contact_plot, create_and_save_cum_contact_prob_plot
```

```python
data_filename = Path(r'C:\Users\martijnsparnaa\surfdrive\PhD\Activity scheduler\Simulations\2021_12\transfer\results_20211215\cum_contact_data.npz')
processed_data_filename = Path(r'C:\Users\martijnsparnaa\surfdrive\PhD\Activity scheduler\Simulations\2021_12\transfer\results_20211215\cum_contact_data_processed.npz')
```

```python
infra_names = [
    'base_case',
    'busy_corner',
    'long_restaurant']

infra_names_2_label = {
    'base_case': 'Base case',
    'busy_corner': 'Busy corner', 
    'long_restaurant': 'Long restaurant'}

measures = (
    'nn_nts_bvd',
    'n_nts_bvd',
    'nn_ts_bvd',
    'nn_nts_svd',
    'nn_nts_lvd'    
)

measures_2_labels = {
    'nn_nts_bvd': 'Base',
    'n_nts_bvd': 'Neighbourhoods',
    'nn_ts_bvd': 'Time slots',
    'nn_nts_svd': 'Shorter visits',
    'nn_nts_lvd': 'Longer visits'
}

comb_sizes = (1,2,3)
```

```python
divider_config = {
    'left':0.62,  
    'gap_hor':0.0,
    'right':0.05,
    'bottom':0.88,
    'gap_ver':0.0,
    'top':0.40,
    'v_legend':0.42,
    'share_y':True
}

main_ax_config = {'grid':{'visible':True, 'axis':'y'},
                  'x_tick_labels':[measures_2_labels[measure] for measure in measures], 
                  'x_tick_labels_config':{'rotation':45, 'ha':'right'}}

flier_props = {'markersize':4, 'marker':'x', 'markeredgecolor':'0.2',
                  'linewidth':1, 'markeredgewidth':1}

box_plot_config = {'vert':True, 'notch':True, 'patch_artist':True, 'widths':0.8, 'flierprops':flier_props,
                   'whiskerprops':{'linewidth':2},'capprops':{'linewidth':2},
                    'boxprops':{'linewidth':2, 'fill':False}, 'medianprops':{'linewidth':2, 'color':'tab:orange'}}

violin_plot_config = {'widths':0.8, 'showmeans':False, 'showextrema':False, 'showmedians':False}
violin_body_plot_config = {'alpha':0.7}

viridis = mpl.colormaps['viridis']
violin_color_order = [viridis(0.5), viridis(0.9), viridis(0.1)]


line_color_order = [viridis(0.5), viridis(0.9), viridis(0.3), viridis(0.7), viridis(0.1)]
style_order = ['-', '--', ':', '-.', '-']
width_order = [1.5]*9#[3, 2.5, 2, 1.5, 3, 2.5, 2, 1.5]
marker_order = ['s', 'X', 'o', '*', 'v', '^', '<', '>', 'x']

legend_config = {'loc':'upper center', 'bbox_to_anchor': (0.5, 1.1), 'ncol':5, 'fontsize':9, 
                 'handlelength':1.2, 'handleheight':0.5, 'columnspacing':1.0, 'handletextpad':0.6}

v_line_config = {'linewidth':1.5, 'color':'0.6', 'linestyle':'--', 'zorder':2}

base_plot_config = {'measure_names':measures, 'main_ax_config':main_ax_config, 'v_line_config':v_line_config, 
                    'box_plot_config':box_plot_config, 'violin_plot_config':violin_plot_config, 
                    'violin_body_plot_config':violin_body_plot_config, 'legend_config':legend_config, 
                    'violin_color_order':violin_color_order, 'infra_names':infra_names, 'infra_names_2_label':infra_names_2_label,
                    'style_order':style_order, 'width_order':width_order, 'marker_order':marker_order, 
                    'line_color_order':line_color_order, 'line_config':{'zorder':3}, 'measure_2_label':measures_2_labels}

fig_config_base = {'figsize':(6.5, 3.5)}

save_pd = Path(r'C:\Users\martijnsparnaa\surfdrive\PhD\Graphics')
```

```python
%matplotlib wx

plt.style.use('tex_image_style.mplstyle')

interaction_type = 'customer_customer'

save_filename = save_pd.joinpath(f'{interaction_type}_cum_contact_times.pdf')

fig_config = deepcopy(fig_config_base)
fig_config.update({'title': f'Cum. contact probs - {interaction_type}'})

plot_config = deepcopy(base_plot_config)
plot_config['interaction_type'] = interaction_type

plots_config = []

for ii, comb_size in enumerate((1,2,3)):
    plot_config_sub = deepcopy(plot_config)
    plot_config_sub['comb_size'] = comb_size
    sub_groups = []
    for infra_name in infra_names:
        sub_group = {'infra_name':infra_name,
                    'label': infra_names_2_label[infra_name]}
        sub_groups.append(sub_group)
    plot_config_sub['sub_groups'] = sub_groups
    plot_config_sub['main_ax_config']['title'] = f'({chr(97+ii)}) N = {comb_size}'
    if ii == 0:
        plot_config_sub['main_ax_config']['y_label'] = 'Contact time [min]'
    else:
        plot_config_sub['main_ax_config']['y_ticks_visible'] = False
        
    plots_config.append(plot_config_sub)
    
create_and_save_cum_contact_plot(fig_config, divider_config, plots_config, data_filename, save_filename)
```
