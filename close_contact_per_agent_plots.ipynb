```python
from pathlib import Path
from copy import deepcopy

import matplotlib.pyplot as plt
import matplotlib as mpl

from contact_plots import create_and_save_abs_close_contact_plot, get_data_from_files, create_and_save_abs_close_contact_bar_plot
```

```python
data_filename = Path(r'C:\Users\martijnsparnaa\surfdrive\PhD\Activity scheduler\Simulations\2021_12\transfer\results_20211215\contact_data.npz')

data_folder = Path(r'C:\Users\martijnsparnaa\surfdrive\PhD\Activity scheduler\Simulations\2021_12\transfer\results_20211215')

contact_data_list, infra_and_measure_name_2_data = get_data_from_files(data_folder, convert_2_minutes=True)
```

```python
infra_names = [
    'base_case',
    'busy_corner',
    'long_restaurant']

infra_names_2_label = {
    'base_case': 'Base case',
    'busy_corner': 'Busy corner', 
    'long_restaurant': 'Long restaurant'}

measures = (
    'nn_nts_bvd',
    'n_nts_bvd',
    'nn_ts_bvd',
    'nn_nts_svd',
    'nn_nts_lvd'    
)

interaction_type_labels = {'customer_customer':'Customer - Customer', 
                           'customer_staff':'Customer - Staff', 
                           'staff_customer':'Staff - Customer'}

measures_2_labels = {
    'nn_nts_bvd': 'Base',
    'n_nts_bvd': 'Neighbourhoods',
    'nn_ts_bvd': 'Time slots',
    'nn_nts_svd': 'Shorter visits',
    'nn_nts_lvd': 'Longer visits'
}
```

```python
divider_config = {
    'left':0.65,  
    'gap_hor':0.8,
    'right':0.1,
    'bottom':1.00,
    'gap_ver':0.3,
    'top':0.0,
    'v_legend':0.50,
    'share_y':False
}

main_ax_config = {'grid':{'visible':True, 'axis':'y'}, 'y_label_config':{'fontsize':10},
                  'y_tick_labels_config':{'labelsize':9}, 
                  'x_tick_labels':[measures_2_labels[measure] for measure in measures], 
                  'x_tick_labels_config':{'fontsize':10, 'rotation':45, 'ha':'right'}, 
                  'title_config':{'fontsize':10}}

flier_props = {'markersize':4, 'marker':'x', 'markeredgecolor':'0.2',
                  'linewidth':1, 'markeredgewidth':1}

box_plot_config = {'vert':True, 'notch':True, 'patch_artist':True, 'widths':0.8, 'flierprops':flier_props,
                   'whiskerprops':{'linewidth':2},'capprops':{'linewidth':2},
                    'boxprops':{'linewidth':2, 'fill':False}, 'medianprops':{'linewidth':2, 'color':'tab:orange'}}#, 'facecolor':'tab:blue'}}

bar_plot_config = {'width':0.8}

violin_plot_config = {'widths':0.8, 'showmeans':False, 'showextrema':False, 'showmedians':False}
violin_body_plot_config = {'alpha':0.7}

viridis = mpl.colormaps['viridis']

violin_color_order = [viridis(0.5), viridis(0.9), viridis(0.1)]

bar_color_order = [
    [viridis(0.5), viridis(0.40), viridis(0.6)],
    [viridis(0.85), viridis(0.75), viridis(0.95)],
    [viridis(0.15), viridis(0.05), viridis(0.25)]
]

legend_config = {'loc':'upper center', 'ncol':3, 'title':'Criterium', 'fontsize':9, 'handlelength':1.2, 
                          'handleheight':0.5, 'title_fontsize':9, 'columnspacing':1.0, 'handletextpad':0.6}

v_line_config = {'linewidth':1.5, 'color':'0.6', 'linestyle':'--', 'zorder':2}

base_plot_config = {'measure_names':measures, 'main_ax_config':main_ax_config, 'v_line_config':v_line_config, 
                    'box_plot_config':box_plot_config, 'violin_plot_config':violin_plot_config, 
                    'violin_body_plot_config':violin_body_plot_config, 'bar_plot_config':bar_plot_config,
                    'legend_config':legend_config, 'violin_color_order':violin_color_order, 'bar_color_order':bar_color_order}

fig_config_base = {'figsize':(6.5, 4)}

save_pd = Path(r'C:\Users\martijnsparnaa\surfdrive\PhD\Graphics')
```

```python
%matplotlib wx

interaction_type = 'staff_customer'

save_filename = save_pd.joinpath(f'{interaction_type}_abs_contact_counts.pdf')

fig_config = deepcopy(fig_config_base)
fig_config.update({'title': f'Contact distributions - {interaction_type}'})

plot_config = deepcopy(base_plot_config)
plot_config['interaction_type'] = interaction_type

plots_config = []

plot_config_sub = deepcopy(plot_config)
plot_config_sub['t_label'] = 'nc_dist=1.0_time=1.0_c_dist=1.5_time=15.0'
sub_groups = []
for infra_name in infra_names:
    sub_group = {'infra_name':infra_name,
                'label': infra_names_2_label[infra_name]}
    sub_groups.append(sub_group)
plot_config_sub['sub_groups'] = sub_groups
plot_config_sub['main_ax_config']['title'] = '(a) Close contact distribution over agents'
plot_config_sub['main_ax_config']['y_label'] = '# Close contacts per agent [-]'
plots_config.append(plot_config_sub)

plot_config_sub = deepcopy(plot_config)
plot_config_sub['data_name'] = '1.5_staff_customer_contacts_over_agents'
sub_groups = []
for infra_name in infra_names:
    sub_group = {'infra_name':infra_name,
                'label': infra_names_2_label[infra_name]}
    sub_groups.append(sub_group)
plot_config_sub['sub_groups'] = sub_groups
plot_config_sub['main_ax_config']['title'] = '(b) Contacts witin 1.5m distribution over agents'
plot_config_sub['main_ax_config']['y_label'] = '# Contacts per agent [-]'
plots_config.append(plot_config_sub)

create_and_save_abs_close_contact_plot(fig_config, divider_config, plots_config, data_filename, save_filename, infra_and_measure_name_2_data)
```
